```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = "/content/drive/MyDrive/Visualization/Covid19Mobility/Region_Mobility_Report_CSVs/2020_GB_Region_Mobility_Report.csv"
df_2020 = pd.read_csv(file_path)

uk_2020 = df_2020[df_2020["sub_region_1"].isna()].copy()
uk_2020["date"] = pd.to_datetime(uk_2020["date"])
uk_2020 = uk_2020.sort_values("date")

# Lockdown start date (UK first national lockdown)
lockdown_date = pd.to_datetime("2020-03-23")
```

```python
# 3.2 (A) Line chart: Workplace mobility over time (with lockdown marker)
plt.figure(figsize=(12, 6))

plt.plot(
    uk_2020["date"],
    uk_2020["workplaces_percent_change_from_baseline"],
)

plt.axvline(lockdown_date, linestyle="--")
plt.title("UK Mobility Trend (2020): Workplaces")
plt.xlabel("Date")
plt.ylabel("Percent change from baseline")
plt.grid(True)

# Save for report
plt.tight_layout()
plt.savefig("fig_workplaces_line_2020.png", dpi=300)
plt.show()
```

```python
# 3.2 (B) Line chart: Residential mobility over time (with lockdown marker
plt.figure(figsize=(12, 6))

plt.plot(
    uk_2020["date"],
    uk_2020["residential_percent_change_from_baseline"],
)

plt.axvline(lockdown_date, linestyle="--")
plt.title("UK Mobility Trend (2020): Residential")
plt.xlabel("Date")
plt.ylabel("Percent change from baseline")
plt.grid(True)

plt.tight_layout()
plt.savefig("fig_residential_line_2020.png", dpi=300)
plt.show()
```

```python
# 3.2 (C) Multi-series line chart: Compare key categories
plt.figure(figsize=(12, 6))

categories = {
    "Workplaces": "workplaces_percent_change_from_baseline",
    "Retail & recreation": "retail_and_recreation_percent_change_from_baseline",
    "Transit stations": "transit_stations_percent_change_from_baseline",
    "Residential": "residential_percent_change_from_baseline",
}

for label, col in categories.items():
    plt.plot(uk_2020["date"], uk_2020[col], label=label)

plt.axvline(lockdown_date, linestyle="--")
plt.title("UK Mobility Trends (2020): Key Categories")
plt.xlabel("Date")
plt.ylabel("Percent change from baseline")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.savefig("fig_multiline_key_categories_2020.png", dpi=300)
plt.show()
```

```python
# 3.2 (D) Box plot: Pre-lockdown vs Lockdown (Workplaces + Residential)
# This creates a clean before vs during comparison
# Define pre vs during lockdown periods
pre = uk_2020[uk_2020["date"] < lockdown_date]
during = uk_2020[uk_2020["date"] >= lockdown_date]

plt.figure(figsize=(10, 6))

data = [
    pre["workplaces_percent_change_from_baseline"].dropna(),
    during["workplaces_percent_change_from_baseline"].dropna(),
    pre["residential_percent_change_from_baseline"].dropna(),
    during["residential_percent_change_from_baseline"].dropna(),
]

labels = [
    "Workplaces (Pre)",
    "Workplaces (Lockdown)",
    "Residential (Pre)",
    "Residential (Lockdown)",
]

plt.boxplot(data, tick_labels=labels)
plt.title("UK Mobility (2020): Pre-lockdown vs Lockdown Comparison")
plt.ylabel("Percent change from baseline")
plt.grid(True)

plt.tight_layout()
plt.savefig("fig_boxplot_pre_vs_lockdown_2020.png", dpi=300)
plt.show()
```

```python
# This is a “visual summary” (not mandatory), using monthly averages:
tmp = uk_2020.copy()
tmp["month"] = tmp["date"].dt.to_period("M").astype(str)

monthly = tmp.groupby("month")[[
    "workplaces_percent_change_from_baseline",
    "retail_and_recreation_percent_change_from_baseline",
    "transit_stations_percent_change_from_baseline",
    "residential_percent_change_from_baseline"
]].mean()

plt.figure(figsize=(10, 4))
plt.imshow(monthly.values, aspect="auto")
plt.yticks(range(len(monthly.index)), monthly.index)
plt.xticks(range(len(monthly.columns)), monthly.columns, rotation=45, ha="right")
plt.title("UK Mobility (2020): Monthly Average Changes (Heatmap View)")
plt.tight_layout()
plt.savefig("fig_heatmap_monthly_2020.png", dpi=300)
plt.show()
```
